1.Q1 Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related use cases.

- A Dashboard is a collection of several worksheets and supporting information displayed in a single place, allowing you to monitor and compare a variety of data simultaneously.
A Story Point is an individual sheet within a Tableau Story, which is a sequence of visualizations (worksheets or dashboards) arranged to work together to convey a specific narrative.

Use Cases :-

Operational Efficiency Dashboard:Healthcare administrators can use dashboards to monitor real-time metrics such as bed occupancy, patient flow through the Emergency Department, and staffing levels. This allows them to identify and resolve bottlenecks immediately to ensure smooth facility operations.

Patient Outcome Story Point Narrative:A story can be used to present a "Patient First" quality improvement project. Individual story points might guide stakeholders through a sequence showing:An initial baseline of readmission rates for heart attack patients.The implementation of new physician schedules.The final outcome, such as a 12% reduction in readmissions and improved patient satisfaction scores.

2. Q2 What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.

- Actions are interactive features that allow users to interact directly with data in a dashboard or worksheet. Instead of using static filters, an action enables a user to click, hover over, or select a data point in one view to trigger a response (like filtering, highlighting, or navigating) in another.

Filter Actions for Hospital Management:---
Drill-Down Analysis:
A manager can click on a specific "Department" (e.g., Cardiology) in a high-level overview chart. The filter action then instantly updates a detailed list to show only the patients, wait times, and staff availability for that specific department.
Identifying Outliers:
If a heat map shows a spike in "Readmission Rates" for a certain age group, clicking that group can filter a secondary dashboard to reveal the specific diagnoses or doctors associated with those patients.
Resource Allocation: By selecting a specific time block on a "Patient Intake" chart, management can see exactly which facilities were over-capacity during that window, helping them adjust staffing for future shifts.
Patient Journey Mapping: Selecting a unique Patient ID in one view can filter across various sheets to show that patient’s entire history, from ER admission to lab results and final discharge, providing a holistic view of care.
